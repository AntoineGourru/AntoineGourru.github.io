```python
def safe_parse_date(entry):
    for key in ["published", "updated"]:
        value = entry.get(key)
        if value:
            try:
                dt = dateparser.parse(value)
                # Si la date n'a pas de timezone, on la met en UTC par défaut
                if dt.tzinfo is None:
                    dt = dt.replace(tzinfo=timezone.utc)
                return dt
            except Exception:
                pass
    return None
```

```python
def strip_html(text):
    if not text:
        return ""
    # retire les tags HTML de manière simple (OK pour un TD)
    text = re.sub(r"<.*?>", " ", text)
    text = re.sub(r"\s+", " ", text).strip()
    return text

def fetch_feed(url, source_name=None):
    d = feedparser.parse(url)
    feed_title = d.feed.get("title", "") if hasattr(d, "feed") else ""
    source = source_name or feed_title or url

    rows = []
    for e in getattr(d, "entries", []):
        title = (e.get("title") or "").strip()
        link = (e.get("link") or "").strip()
        summary = e.get("summary") or e.get("description") or ""
        summary = strip_html(summary)
        dt = safe_parse_date(e)

        rows.append({
            "source": source,
            "feed_url": url,
            "title": title,
            "url": link,
            "summary": summary,
            "published_at": dt
        })
    return rows
```

```python
def filter_by_keywords(df, keywords):
    if not keywords:
        return df.copy()
    pattern = "|".join(re.escape(k) for k in keywords)
    mask = (
        df["title"].str.contains(pattern, case=False, na=False) |
        df["summary"].str.contains(pattern, case=False, na=False)
    )
    return df[mask].copy()

keywords = ["intelligence artificielle"]
df_ai = filter_by_keywords(dfn, keywords)
df_ai[["published_at", "source", "title", "url"]].head(20)
```

```python
def add_score(df, keywords):
    df = df.copy()
    if not keywords:
        df["score"] = 0
        return df

    pattern = "|".join(re.escape(k) for k in keywords)
    # Compter occurrences (titre pondéré x2)
    title_counts = df["title"].str.count(pattern, flags=re.IGNORECASE)
    sum_counts = df["summary"].str.count(pattern, flags=re.IGNORECASE)
    df["score"] = 2 * title_counts + sum_counts
    return df

df_scored = add_score(df_ai_jan, keywords)
df_scored.sort_values(["score", "published_at"], ascending=[False, False]).head(20)[
    ["score", "published_at", "source", "title", "url"]
]
```
